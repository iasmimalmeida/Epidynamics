```python
import numpy as np
import pandas as pd 
from scipy import stats
import matplotlib.pyplot as plt
```

```python
update_prof = {'Episódico/Epidêmico':'Episodic/Epidemic', 'Epidêmico':'Epidemic', 'Episódico':'Episodic',
       'Transmissão Persistente':'Persistent Transmission', 'Sem transmissão': 'No Transmission', 
              'Sem transmissao': 'No Transmission'}
```

```python
df_10_19 = pd.read_csv('../Determinantes/clima/clima_agg_10_19.csv', index_col = 'Unnamed: 0')

df_10_19.loc[:, 'dengue_pattern'] = df_10_19.dengue_pattern.replace(update_prof)

df_10_19['period'] = '10_19'

df_10_19.head()
```

```python
df_20_22 = pd.read_csv('../Determinantes/clima/clima_agg_20_22.csv', index_col = 'Unnamed: 0')

df_20_22.loc[:, 'dengue_pattern'] = df_20_22.dengue_pattern.replace(update_prof)

df_20_22['period'] = '20_22'

df_20_22.head()
```

```python
col = 'temp_med'
perfil = 'No Transmission'

df_10_19_ = df_10_19.loc[df_10_19.dengue_pattern == perfil]
df_20_22_ = df_20_22.loc[df_20_22.dengue_pattern == perfil]
```

```python
stats.ttest_ind(df_10_19_[col], df_20_22_[col])
```

```python
col = 'temp_med'
perfil = 'Persistent Transmission'

df_10_19_ = df_10_19.loc[df_10_19.dengue_pattern == perfil]
df_20_22_ = df_20_22.loc[df_20_22.dengue_pattern == perfil]
```

```python
stats.ttest_ind(df_10_19_[col], df_20_22_[col]).pvalue
```

```python
df_20_22.dengue_pattern.unique()
```

```python

df_p = pd.DataFrame()
for col in ['temp_med', 'umid_med', 'precip_tot', 'thr_temp',
       'thr_umid', 'thr_prec', 'thr_temp_umid']:
    p_values = np.array([])
    for perfil in ['Episodic/Epidemic', 'Epidemic', 'Episodic',
       'Persistent Transmission', 'No Transmission']:
        #print(perfil)

        df_10_19_ = df_10_19.loc[df_10_19.dengue_pattern == perfil]
        df_20_22_ = df_20_22.loc[df_20_22.dengue_pattern == perfil]

        p_values = np.append(p_values, 
                            stats.ttest_ind(df_10_19_[col], df_20_22_[col]).pvalue)

    df_p_ = pd.DataFrame([p_values], columns = ['Episodic/Epidemic', 'Epidemic', 'Episodic',
       'Persistent Transmission', 'No Transmission'], index = [col])

    df_p = pd.concat([df_p, df_p_])

df_p 
```

```python
def color_cells(val):
    color = 'background-color: red' if val > 0.05 else ''
    return color

# Apply the function to the DataFrame to color the cells
df_p.style.applymap(color_cells)
```
